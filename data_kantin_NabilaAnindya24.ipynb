```python
#Latihan 1: Operasi Dasar NumPy Array

import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

```python
#Latihan 2

import pandas as pd

data_kantin = {
 'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
 'harga': [12000, 4000, 10000, 4000, 8000],
 'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Tugas Analisis 2:

Amati tabel yang dihasilkan. Kolom mana yang terlihat memiliki data kosong  (None)? Menurutmu apa risikonya jika data ini langsung dianalisis tanpa dibersihkan?

= Ada 2 kolom yang menghasilkan data kosong yaitu menu (baris ke 5), terjual (baris ke 3). Resikonya perhitungan dan analisis jadi tidak lengkap dan tidak akurat.


```python
#LATIHAN 3
df = pd.read_csv('data_kantin.csv')

print(df.head()) #5 baris pertama
print(df.info()) # tipe data dan jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)


```

**Tugas Analisis 3:**
Dari hasil df.info(), kolom mana yang jumlah non-null-nya lebih sedikit dari  jumlah baris total? Apa artinya?    

= Baris total nya ada 110 dan di bagian menu (107  non-null), terjual (104 non-null).
Artinya, kolom menu kekurangan 3 data dan kolom terjual kekurangan 6 data. Data yang hilang ini adalah missing value (NaN/None).

```python
#Latihan 4: Menangani Missing Value

print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisis 4:

Mengapa pada contoh di atas kolom terjual diisi (fillna) sedangkan baris  dengan menu kosong justru dihapus (dropna)? Diskusikan alasan logisnya.

= Karena jumlah_terjual adalah data yang isinya angka, jadi masih bisa ditebak nilainya.
Fillna itu masih bisa diperkirakan dan baris lainnya masih berguna, sementara dropna itu gabisa ditebak jadi ga berguna kalau di pertahankan.


```python
#Latihan 5: Menangani Duplikat dan Tipe Data

print(df.duplicated().sum()) # jumlah baris duplikat

df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer

print(df.dtypes)

print("Jumlah baris sebelum:", df.shape[0])
df = df.drop_duplicates()
print("Jumlah baris sesudah:", df.shape[0])
```

Tugas Analisis 5:

Setelah drop_duplicates(), berapa jumlah baris dataset sekarang dibanding  sebelumnya? Mengapa penting memastikan tipe data (dtypes) sudah benar sebelum
data dianalisis lebih lanjut?

= awalnya ada 5 baris duplikat di dataset, duplikast itu di hapus pakai drop_duplicates() makanya pas di cek ulang di akhir, jumlah baris sebelum dan sesudah sama sama 102- itu karena duplikatnya sudah terlanjur dihapus duluan, bukan karena memang dari awal nggak ada duplikat.

```python
#Latihan 6: Data Manipulation (Filtering, Sorting, Groupby)

laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
